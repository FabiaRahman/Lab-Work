```python
# NON PRE EMPTIVE SJF

processes = [
    {"PID": "P1", "AT": 3, "BT": 1},
    {"PID": "P2", "AT": 2, "BT": 4},
    {"PID": "P3", "AT": 3, "BT": 6},
    {"PID": "P4", "AT": 1, "BT": 3},
    {"PID": "P5", "AT": 4, "BT": 2},
    {"PID": "P6", "AT": 6, "BT": 1},
]

time = 0
completed = []
total_wt = 0
total_tat = 0

def find_shortest(ready):
    shortest = ready[0]

    for p in ready:
        if p["BT"] < shortest["BT"]:
            shortest = p

    return shortest

while len(completed) < len(processes):

    ready = []

    for p in processes:
        if p["AT"] <= time and p not in completed:
            ready.append(p)

    if not ready:
        time += 1
        continue

    p = find_shortest(ready)

    time += p["BT"]

    p["CT"] = time
    p["TAT"] = p["CT"] - p["AT"]
    p["WT"] = p["TAT"] - p["BT"]

    total_wt += p["WT"]
    total_tat += p["TAT"]

    completed.append(p)

print("PID\tAT\tBT\tCT\tTAT\tWT")

for p in completed:
    print(
        p["PID"], "\t",
        p["AT"], "\t",
        p["BT"], "\t",
        p["CT"], "\t",
        p["TAT"], "\t",
        p["WT"]
    )

# Average Waiting Time
average_wt = total_wt / len(processes)
# Average Turnaround Time
average_tat = total_tat / len(processes)

print("\nAverage Waiting Time =", average_wt)
print("Average Turnaround Time =", average_tat)
```

```python
# PRE EMPTIVE SJF

processes = [
    {"PID": "P1", "AT": 3, "BT": 1},
    {"PID": "P2", "AT": 2, "BT": 4},
    {"PID": "P3", "AT": 3, "BT": 6},
    {"PID": "P4", "AT": 1, "BT": 3},
    {"PID": "P5", "AT": 4, "BT": 2},
    {"PID": "P6", "AT": 6, "BT": 1},
]

time = 0
completed = []
total_wt = 0
total_tat = 0

for p in processes:
    p["Remaining"] = p["BT"]


def find_shortest(ready):
    shortest = ready[0]

    for p in ready:
        if p["Remaining"] < shortest["Remaining"]:
            shortest = p

    return shortest


while len(completed) < len(processes):

    ready = []

    for p in processes:
        if p["AT"] <= time and p not in completed and p["Remaining"] > 0:
            ready.append(p)

    if not ready:
        time += 1
        continue

    p = find_shortest(ready)

    p["Remaining"] -= 1
    time += 1

    if p["Remaining"] == 0:
        p["CT"] = time
        p["TAT"] = p["CT"] - p["AT"]
        p["WT"] = p["TAT"] - p["BT"]

        total_wt += p["WT"]
        total_tat += p["TAT"]

        completed.append(p)


print("PID\tAT\tBT\tCT\tTAT\tWT")

for p in completed:
    print(
        p["PID"], "\t",
        p["AT"], "\t",
        p["BT"], "\t",
        p["CT"], "\t",
        p["TAT"], "\t",
        p["WT"]
    )


# Average Waiting Time
average_wt = total_wt / len(processes)

# Average Turnaround Time
average_tat = total_tat / len(processes)

print("\nAverage Waiting Time =", average_wt)
print("Average Turnaround Time =", average_tat)
```

```python
# FIRST COME, FIRST SERVED

processes = [
    {"PID": "P1", "AT": 3, "BT": 1},
    {"PID": "P2", "AT": 2, "BT": 4},
    {"PID": "P3", "AT": 3, "BT": 6},
    {"PID": "P4", "AT": 1, "BT": 3},
    {"PID": "P5", "AT": 4, "BT": 2},
    {"PID": "P6", "AT": 6, "BT": 1}
]

def get_arrival_time(p):
    return p["AT"]

processes.sort(key=get_arrival_time)

time = 0
completed = []
total_wt = 0
total_tat = 0

for p in processes:

    if time < p["AT"]:
        time = p["AT"]

    time += p["BT"]

    p["CT"] = time
    p["TAT"] = p["CT"] - p["AT"]
    p["WT"] = p["TAT"] - p["BT"]

    total_wt += p["WT"]
    total_tat += p["TAT"]

    completed.append(p)

print("PID\tAT\tBT\tCT\tTAT\tWT")

for p in completed:
    print(
        p["PID"], "\t",
        p["AT"], "\t",
        p["BT"], "\t",
        p["CT"], "\t",
        p["TAT"], "\t",
        p["WT"]
    )

average_wt = total_wt / len(processes)
average_tat = total_tat / len(processes)

print("\nAverage Waiting Time =", average_wt)
print("Average Turnaround Time =", average_tat)
```
