```python
#corrected code

#dictionary is used
process_list = [
    {"PID": "P1", "AT": 3, "BT": 3},
    {"PID": "P2", "AT": 2, "BT": 1},
    {"PID": "P3", "AT": 5, "BT": 2},
    {"PID": "P4", "AT": 0, "BT": 3},
    {"PID": "P5", "AT": 1, "BT": 2}
]

def get_arrival_time(item):
    return item["AT"]
#sorting the arrival time
process_list.sort(key=get_arrival_time)

time = 0
total_tat = 0
total_wt = 0

print("PID\tAT\tBT\tCT\tTAT\tWT")

#assigning values accordingly
for i in process_list:
    PID = item["PID"]
    AT = item["AT"]
    BT = item["BT"]

#calculation of completion time, turnaround time and waiting time
    if time < AT:
        time = AT
    time += BT
    CT = time
    TAT = CT - AT
    WT = TAT - BT

    total_tat += TAT
    total_wt += WT

    print(PID, "\t", AT, "\t", BT, "\t",
          CT, "\t", TAT, "\t", WT)

avg_tat = total_tat / len(process_list)
avg_wt = total_wt / len(process_list)

print("\nAverage Turnaround Time:", avg_tat)
print("Average Waiting Time:", avg_wt)
```
